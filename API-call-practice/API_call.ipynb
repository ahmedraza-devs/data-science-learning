## NEW API call (Scrapethissite)

```python
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
res.status_code
```

```python
if res.status_code == 200:
    print(res.headers)

with open("scrapped_data/data1.html" , "w" , encoding="utf-8") as f:
    f.write(res.text)
```

## For try and catch:

```python
try:  
    r = requests.get("https://www.scrapethissite.com//simple/" , timeout=5)  
    r.raise_for_status()   
# Raises error for 4xx/5xx responses  
except requests.exceptions.RequestException as e:  
    print("Error:", e) 
```

## HEaders:

```python
headers = {  
    "User-Agent": "Mozilla/5.0"  
}  
response = requests.get("https://www.scrapethissite.com/simple/", headers=headers) 
print(response.text)
```

# BeautifulSoup...

```python
from bs4 import BeautifulSoup
```

```python
with open("scrapped_data/data1.html" , "r" , encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content , "lxml")
```

```python
all_h3 = soup.find_all("h3")


country_info = [ ]

for h3 in all_h3:
    country_names = h3.get_text(strip = True)   
    country_capitals = h3.find_next("div").select("span.country-capital")[0].get_text(strip = True)
    country_info.append([country_names , country_capitals])
```

```python
import pandas as pd
df = pd.DataFrame(country_info , columns= ["Country Name" , "Country Capital"])
df
```

```python
df.to_csv("AR_cleaning/cleaned_data1.csv" , index=False)
```
