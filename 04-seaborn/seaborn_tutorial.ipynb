# Heatmap:

```python
sns.get_dataset_names()
flights = sns.load_dataset("flights")
```

```python
flights
```

```python
flight_pivot = flights.pivot(index ="month" , columns = "year", values="passengers")
```

```python
sns.heatmap(
    flight_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)

plt.title("Passenger heatmap")
plt.tight_layout()
```

# Best practice to plot:

```python
fig , ax = plt.subplots()

sns.lineplot(
    data= tips,
    x = "day",
    y= "total_bill",
    hue = "sex",
    ax = ax,
    errorbar=None
)

ax.set_title("Avg Bill values")
ax.set_xlabel("Days")
ax.set_ylabel("Bill amount")

fig.tight_layout()
```
