# Data Transformation (Feature Learning).

```python
df2 = df.copy()

# apply function:
df2["tax"] = df2["income"].apply(lambda x: "20%" if x >= 50000 else "10%")
df2

# map func:
gender_map = {"Male" : "M", "Female" : "F" , "Unknown" : "U"}
df2["gender"] = df2["gender"].map(gender_map)
df2

# assign function:
new_income = df2["income"] * 1.1
df2.assign(new_income = new_income)

# Replace (old , new) func:
df2["country"] = df2["country"].replace("USA" , "US")
df2

# Rename func:
df2.rename(columns = {"income" : "Salary"})
df2.rename(index = {4 : "Fourth"})

# Sorting Values:
# sort_value() func:
sorted_df = df2.sort_values(["income", "tax"])

# sort_index func:
df2.sort_index()  # all thing are reset to orginal

# reset_index func:
df2.reset_index() # new index column will be created to reserve the org index
df2.reset_index(drop = True) # here no reservation of the old indexes 

# rank function:
sorted_df["Rankings"] = sorted_df["income"].rank(ascending=False ,method= "dense")
sorted_df
```

# Gouped By & Aggregiate Functions:

```python
df2 = df.copy()

# for group by :
df2.groupby("country")["income"].mean()  # like this max and min and  sum 

# for using multiple functions of mean and sum use aggregiate (agg):
df2.groupby("country")["income"].agg(["mean" , "max" , "min"])
df2.groupby("country")["income"].agg( avg_income = "mean" , max_income =  "max" , min_income =  "min")

# now for group two different thing with two different funcs:
df2.groupby("country").agg({
    "income" : "max",
    "age" : "mean"
})
# name labels:
df2.groupby("country").agg(
    max_income = ("income" , "max"),
    avg_age = ("age" , "mean")
)
```

# MELT & PIVOT

```python
df3 = pd.DataFrame({
    "country": ["USA", "USA", "UK", "Canada", "UK"],
    "year": [2022, 2023, 2022, 2023, 2023],
    "sales": [1200, 1500, 900, 1100, 1300],
    "profit": [250, 320, 180, 220, 270]
})
df3
```

```python
melted_data = df3.melt(
    id_vars= ["country" , "year"],
    value_vars= ["sales", "profit"],
    var_name= "matrics",
    value_name= "value"
)
melted_data

original = melted_data.pivot(
    index=["country" , "year"],
    columns="matrics",
    values="value"
)
original
```

# Merge and Join

```python
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 5],
    'name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Bilal'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'Marketing']
})

salaries = pd.DataFrame({
    'emp_id': [1, 2, 3, 6],
    'salary': [70000, 65000, 80000, 90000],
    'bonus': [5000, 4000, 6000, 7000]
})

# joins:
pd.merge(employees , salaries , on="emp_id" , how="inner")
pd.merge(employees , salaries , on="emp_id" , how="left")
pd.merge(employees , salaries , on="emp_id" , how="right")
pd.merge(employees , salaries , on="emp_id" , how="outer")
```

# Concatenation

```python
sales_q1 = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard"],
    "sales": [120, 300, 150]
})

sales_q2 = pd.DataFrame({
    "product": ["Monitor", "Printer", "Speaker"],
    "sales": [80, 60, 140]
})

pd.concat([sales_q1 , sales_q2] , ignore_index=True)   
pd.concat([sales_q1 , sales_q2] , axis=1)
```
